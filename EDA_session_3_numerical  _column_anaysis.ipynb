**capping**
- we can fill the outliers which is less than 25pvalues
- we can fill outlier which is less than 75p with 75pvalue

```python
wage_data=visa_df['prevailing_wage']

q1 = np.percentile(wage_data, 25)
q2 = np.percentile(wage_data, 50)
q3 = np.percentile(wage_data, 75)

IQR = q3 - q1

LB = q1 - 1.5 * IQR
UB = q3 + 1.5 * IQR
visa_df['wages_data_cap']=wage_data.clip(LB,UB)
```

```python
wage_data=visa_df['prevailing_wage']
wage_data_q2=visa_df['wages_q2']
wage_data_cap=visa_df['wages_data_cap']
plt.figure(figsize=(10,4))
plt.subplot(1,2,1).boxplot(wage_data,vert=False)
plt.subplot(1,2,2).boxplot(wage_data_q2,vert=False)
plt.subplot(1,2,2).boxplot(wage_data_cap,vert=False)

plt.show()
```
