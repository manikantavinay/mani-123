**KNNimputer**

- KNN : K-nearest neighbours
- where k is a number 
- instaed of taking avg of all  the values data
    - first will find neighbours who are closed with missing data
    - by using Euclidian distance  calculation  will idenify the neighbour rqwa
    - suppose k=2 will idenify the two closed  rows
    - those closest rows  values will take it,will calculation the mean
    - finally will fill with the mean

```python
salary=[50000,np.nan,70000,90000]
age=[25,28,30,40]
df=pd.DataFrame(zip(salary,age),columns=['salary','Age'])
```

```python
df
```

- sklearn

```python
from sklearn.impute import KNNImputer
imputer=KNNImputer(n_neighbors=20)
pd.DataFrame(imputer.fit_transform(df))
```
