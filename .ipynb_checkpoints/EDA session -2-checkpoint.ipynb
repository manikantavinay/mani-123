```python
**Bar chart**
-  X- axis :class names(keys)
- y-axis: Frequency(values)
```

```python
keys=visa_df['continent'].value_counts().keys()

vlues=visa_df['continent'].value_counts().values
```

```python
from matplotlib.colors import CSS4_COLORS
```

```python
colors = ['red', 'blue', 'green', 'yellow', 'pink', 'purple']

plt.figure(figsize=(10,6))
plt.bar(keys, vlues, color=colors)

plt.xlabel('Continents')
plt.ylabel('No Of Applicants')
plt.title('Bar Chart')

plt.show()
print(plt.xlabel)
```

**pie chart**
- autopct=%a.bf%%
- first % is the formatting
- a: how much min width
- b:after point how many decimals
- %%: % symbol
- %1.1f%%:40.2%
- %2.2f%%:43.23

```python
keys=visa_df['continent'].value_counts().keys()

vlues=visa_df['continent'].value_counts().values
plt.pie(vlues,explode=[0.1 for i in range(6)],labels=keys,autopct='%1.2f%%',radius=0.8,startangle=90)
plt.savefig('continet_pie.jpg')
plt.show()
```
